## 0. Charger la séquence

Exécutez la cellule ci-dessous : elle crée un fichier FASTA et range la séquence dans une

````python
#@title ▶️ Exécutez cette cellule pour charger la séquence (ne pas modifier)
# Séquence codante de référence (RefSeq) du gène HBB, de l'ATG au codon stop.

FASTA_HBB = """>HBB bêta-globine — NM_000518.5
ATGGTGCATCTGACTCCTGAGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTGAAC
GTGGATGAAGTTGGTGGTGAGGCCCTGGGCAGGCTGCTGGTGGTCTACCCTTGGACCCAG
AGGTTCTTTGAGTCCTTTGGGGATCTGTCCACTCCTGATGCTGTTATGGGCAACCCTAAG
GTGAAGGCTCATGGCAAGAAAGTGCTCGGTGCCTTTAGTGATGGCCTGGCTCACCTGGAC
AACCTCAAGGGCACCTTTGCCACACTGAGTGAGCTGCACTGTGACAAGCTGCACGTGGAT
CCTGAGAACTTCAGGCTCCTGGGCAACGTGCTGGTCTGTGTGCTGGCCCATCACTTTGGC
AAAGAATTCACCCCACCAGTGCAGGCTGCCTATCAGAAAGTGGTGGCTGGTGTGGCTAAT
GCCCTGGCCCACAAGTATCACTAA
"""

with open("hbb.fasta", "w") as f:
    f.write(FASTA_HBB)


def lire_fasta(nom_fichier):
    """Lit un fichier FASTA : ignore la ligne d'en-tête (>...) et recolle les lignes de séquence."""
    sequence = ""
    with open(nom_fichier) as f:
        for ligne in f:
            ligne = ligne.strip()
            if ligne and not ligne.startswith(">"):
                sequence = sequence + ligne
    return sequence


adn = lire_fasta("hbb.fasta")
print("HBB chargé ✅")
````

Le format **FASTA** : une ligne d'en-tête qui commence par `>`, puis la séquence, coupée en
lignes de 60 lettres.

La fonction `lire_fasta()` de la cellule précédente recolle ces lignes : `adn` contient toute
la séquence en **une seule chaîne de caractères**. C'est cette forme qu'on manipulera dans tout

**La méthionine initiale.** Le codon start `AUG` code une méthionine : toute protéine est
traduite à partir d'elle. Dans le globule rouge, elle est ensuite retirée de la bêta-globine :
la protéine mature compte 146 acides aminés, et commence par la valine. La
littérature et les fichiers de structure numérotent les acides aminés à partir de cette valine.

La suite : ces 147 lettres forment une chaîne, qui se replie dans l'eau en une forme
précise. C'est cette forme — et ce qui change chez le patient — que le notebook 2 examine.
